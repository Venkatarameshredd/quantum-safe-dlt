## 2. NIST Post-Quantum Cryptography (ML-DSA / Crystals-Dilithium)

```python
pqc = get_pqc_engine()
pub, priv = pqc.generate_keypair()

sample_deed = b"AP_BHUDHAAR_CADASTRAL_RECORD_VILLAGE_NAVULURU_SURVEY_244_1A"
sig = pqc.sign(sample_deed, priv, pub)
verification = pqc.verify(sample_deed, sig)

print(f"Algorithm: {sig.algorithm}")
print(f"Signature Hex: {sig.signature_hex[:48]}...")
print(f"Is Valid: {verification.is_valid} ({verification.details})")
print(f"Verification Latency: {verification.latency_ms:.2f} ms")
```
